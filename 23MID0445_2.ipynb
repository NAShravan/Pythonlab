```python
#TASK1
import threading
import time

def do_something(seconds):
    print (f'sleeping {seconds} second (s)')
    time.sleep(seconds)
    print('Done sleeping...')

t1 = threading.Thread(target = do_something, args = [2])
t1.start()
t1.join()
```

```python
#TASK3
import threading
import time
start=time.perf_counter()

def do_something(seconds):
    print (f'sleeping {seconds} second (s)')
    time.sleep(seconds)
    print('Done sleeping...')

t1 = threading.Thread(target = do_something, args = [1.5])
t1.start()
t1.join()

finish=time.perf_counter()
print(f'Finished in {round(finish-start,2)} second(s)')
```

```python
#TASK7
import time
import requests
import concurrent.futures

audio_urls = [
    'https://samplelib.com/mp3/sample-3s.mp3',
    'https://samplelib.com/mp3/sample-9s.mp3',
    'https://samplelib.com/mp3/sample-speech-30m.mp3'
]

t1 = time.perf_counter()

def download_audio(audio_url):
    audio_name = audio_url.split('/')[-1]
    with requests.get(audio_url, stream=True) as response:
        with open(audio_name, 'wb') as audio_file:
            for chunk in response.iter_content(chunk_size=8192):
                if chunk:
                    audio_file.write(chunk)
    print(f'{audio_name} was downloaded...')

with concurrent.futures.ThreadPoolExecutor() as executor:
    executor.map(download_audio, audio_urls)

t2 = time.perf_counter()
print(f'Finished in {t2-t1} seconds')
```
